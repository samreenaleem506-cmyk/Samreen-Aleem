## Conditional Statements - Decision making in Python !

```python
#program for marksheet
name =input("Enter your name:")
mathematics=eval(input("Enter mathematics marks(out of 100):"))
english=eval(input("Enter english marks(out of 100):"))
physics=eval(input("Enter physics marks(out of 100):"))
chemistry=eval(input("Enter chemistry marks(out of 100):"))
urdu=eval(input("Enter urdu marks(out of 100):"))

obtain=mathematics+english+physics+chemistry+urdu

percentage=(obtain/500)*100

if percentage>=75:
    print(f"Dear{name},You have scored {obtain} marks,percentage is {round(percentage,2)},and your Garde is A+")
elif percentage>=70:
    print(f"Dear{name},You have scored {obtain} marks,percentage is {round(percentage,2)},and your Grade is A")
elif percentage>=65:
    print(f"Dear{name},You have scored {obtain} marks,percentage is {round(percentage,2)},and your Grade is B+")
elif percentage>=60:
    print(f"Dear{name}, You have scored {obtain} marks,percentage is {round(percentage,2)},and your Grade is B")
elif percentage>=55:
    print(f"Dear{name}, You have scored {obtain} marks,percentage is {round(percentage,2)},and your Grade is C")
else:
    print(f"Dear{name}, You have scored {obtain} marks,percentage {round(percentage,2)},and your Garde is F")
    
```
