```python
df['num_character']=df['Text'].apply(len)
```

```python
df
```

```python
df['num_words']=df['Text'].apply(lambda x:len(nltk.word_tokenize(x)))
```

```python
df
```

```python
df['num_sent']=df['Text'].apply(lambda x:len (nltk.sent_tokenize(x)))
```
